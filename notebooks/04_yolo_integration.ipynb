### Imports

```python
import json
import sys
from pathlib import Path

REPO = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(REPO / "src"))
from yolo_eval.evaluate import evaluate, load_overlap_map, remap_exdark_labels_to_coco

WEIGHTS = "yolov8n.pt"  # COCO-pretrained, frozen zero-shot detector
print("repo:", REPO)
```

### ExDark val baseline

```python
coco_yaml = remap_exdark_labels_to_coco(REPO / "exdark_yolo", load_overlap_map())
exdark_base = evaluate(WEIGHTS, str(coco_yaml))
print(json.dumps({k: v for k, v in exdark_base.items() if k != "per_class"}, indent=2))
```

### LoLI-Street val baseline

```python
loli_base = evaluate(WEIGHTS, str(REPO / "loli_yolo" / "loli.yaml"))
print(json.dumps({k: v for k, v in loli_base.items() if k != "per_class"}, indent=2))
```
